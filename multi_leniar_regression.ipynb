# Multiple Leniar Regression

```python
import numpy as np
import random
```

##### X1 independent data 1

```python
x1 = np.array([1,2,3,4])
```

##### x2 independant data 2

```python
x2 = np.array([4,5,8,2])
```

##### y depedant data

```python
y = np.array([1,6,8,12]).T
```

##### *y = f( x1 , x2)*
##### *y = a0 + a1x1 + a2x2*
### General Equation
##### *y = a0 + a1x1 + a2x2 + ... + AnXn*

```python
X = np.array([np.ones(4),x1,x2],dtype = np.int64).T
```

#### a0 intercept

```python
a0 = random.randint(1,10)
```

#### a1 coefficent of x1

```python
a1 = random.randint(1,10)
```

#### a2 coefficent of x2

```python
a2 = random.randint(1,10)
```

a = {a0,
     a1,
      a2}




```python
a = np.array([a0, a1, a2])
```

*a<sup>^</sup> = ((X<sup>T</sup> X)<sup>-1</sup> X<sup>T</sup>)  y*

```python
#(XT X)
v1 = X.T @ X
#(XT X)-1
v1 = np.linalg.inv(v1)
#(XT X)-1 XT
v1 = v1 @ X.T
#((XT X)-1 XT) y
v1 = v1 @ y
#a^ = ((XT X)-1 XT) y
a = v1
print(f"[a0: {a[0]}|\n|a1: {a[1]} |\n|a2: {a[2]}]")
```

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Create a figure and a 3D axes
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Scatter plot the original data points
ax.scatter(x1, x2, y, color='blue', label='Original Data Points')

# Create a meshgrid for x1 and x2 to plot the regression plane
x1_surf = np.linspace(x1.min(), x1.max(), 100)
x2_surf = np.linspace(x2.min(), x2.max(), 100)
X1_surf, X2_surf = np.meshgrid(x1_surf, x2_surf)

# Calculate the corresponding y values for the regression plane
# y = a[0] + a[1]*x1 + a[2]*x2
Y_surf = a[0] + (a[1] * X1_surf) + (a[2] * X2_surf)

# Plot the regression plane
ax.plot_surface(X1_surf, X2_surf, Y_surf, color='red', alpha=0.5, label='Regression Plane')

# Set labels and title
ax.set_xlabel('X1')
ax.set_ylabel('X2')
ax.set_zlabel('Y')
ax.set_title('Multiple Linear Regression Chart')

# Due to a known issue with plot_surface and legend, we manually create a legend proxy
from matplotlib.lines import Line2D
blue_dot = Line2D([0], [0], marker='o', color='w', label='Original Data Points',
                          markerfacecolor='blue', markersize=10)
red_surface = Line2D([0], [0], marker='s', color='w', label='Regression Plane',
                             markerfacecolor='red', markersize=10, alpha=0.5)
ax.legend(handles=[blue_dot, red_surface])

plt.show()
```
