This is linear regression model from scratch for building a prediction of sales on certain week. Usually what linear regression do is if we input an x value we get predicted y value which means: x is mapped to y in labled data

```python
import csv
import matplotlib.pyplot as plt
```

#data set

#day

```python
x = [1, 2, 3, 4, 5]
```

#sales

```python
y = [1.2, 1.8, 2.6, 3.2, 3.8]
```

#Apply Linear regression algorythm y = mx+c | y = a0 + a1 * x +e (e) is error
#a1 = ((mean(x*y)-mean(x)*mean(y)) / mean(x^2) - mean(x)^2
#a0 = mean(y) - a1 * mean(x)


# caluclate the mean of a list of data

```python
def mean(value:list) -> float:
    lenght = len(value)
    list_sum = sum(value)
    return list_sum/lenght
```

```python
def multi_list(list1:list,list2:list) -> float:
    if len(list1) == len(list2):
        xy = 0
        for i in range(len(list1)):
            xy += list1[i] * list2[i]
        return xy
    else:
        raise ValueError
```

#Calculate a1

```python
xy = multi_list(x,y)
xy_mean = xy/len(x)
x_mean = mean(x)
y_mean = mean(y)
x2 = [v**2 for v in x]
x2_mean = mean(x2)
```

```python
a1 = (xy_mean - x_mean*y_mean)/(x2_mean - x_mean**2)
```

#Calculate a0

```python
a0 = y_mean - (a1 * x_mean)
```

#day

```python
day = int(input("Enter the day: "))
```

#Predicted outcome

```python
sale = a0 + (a1*day)
```

```python
data_csv = [day,sale]
with open("linear_regression.csv", mode = "a", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(data_csv)
```

#Plot the Linear Regresion

```python
plt.scatter(x, y, label='Original Data')
regression_line_y = [a0 + a1 * val for val in x]
plt.plot(x, regression_line_y, color='red', label=f'Regression Line: y = {a1:.2f}x + {a0:.2f}')
plt.xlabel('Day (x)')
plt.ylabel('Sales (y)')
plt.title('Linear Regression Model')
plt.legend()
plt.grid(True)
plt.show()
```
